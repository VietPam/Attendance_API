```python
from vnstock import Vnstock

stock = Vnstock().stock(symbol='STB', source='KBS')
df = stock.quote.history(start='2024-01-01', end='2026-09-19', interval='1D')
print(df.head())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.plot(df['time'], df['close'])
plt.title('Giá đóng cửa STB')
plt.xlabel('Ngày')
plt.ylabel('Giá (nghìn đồng)')
plt.grid(True)
plt.show()
```
